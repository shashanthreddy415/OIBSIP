## Missing Data Handling

The dataset contains missing values in `gender`, `age`, `purchase_amount`, and `payment_method`.

- **Gender:** Missing gender values will be filled using the **mode**, because gender is a categorical variable and the most frequent category provides a reasonable replacement.
- **Age:** Missing age values will be filled using the **median**. Median is preferred because the dataset contains extreme and invalid age values, making it more robust than the mean.
- **Purchase Amount:** Missing purchase amounts will be filled using the **median** because transaction amounts can contain extreme values, and the median is less affected by outliers.
- **Payment Method:** Missing payment method values will be filled using the **mode**, because this is a categorical variable and the most common payment method is an appropriate replacement.

These choices are made to preserve as much data as possible while reducing the influence of extreme values.

```python
df["age"] = pd.to_numeric(df["age"], errors="coerce")
print(df["age"])
```

```python
median_age = df["age"].median()
print("Median age:", median_age)
```

```python
df["age"] = df["age"].fillna(median_age)
print("Missing age values after imputation:", df["age"].isnull().sum())
```

```python
df["gender"] = df["gender"].replace({
    "M": "Male",
    "m": "Male",
    "male": "Male",
    "F": "Female",
    "female": "Female",
    "FEMALE": "Female"})
print(df["gender"].value_counts(dropna=False))
```

```python
gender_mode = df["gender"].mode()[0]
print("Gender mode:", gender_mode)
```

```python
df["gender"] = df["gender"].fillna(gender_mode)
print("Missing gender values after imputation:", df["gender"].isnull().sum())
```

```python
df["payment_method"] = df["payment_method"].replace({
    "upi": "UPI",
    "credit card": "Credit Card",
    "cash": "Cash"})
print(df["payment_method"].value_counts(dropna=False))
```

```python
payment_mode = df["payment_method"].mode()[0]
print("Payment method mode:", payment_mode)
```

```python
df["payment_method"] = df["payment_method"].fillna(payment_mode)
print("Missing payment method values after imputation:", df["payment_method"].isnull().sum())
```

```python
df["purchase_amount"] = (
    df["purchase_amount"]
    .astype("string")
    .str.replace("₹", "", regex=False)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)
df["purchase_amount"] = pd.to_numeric(
    df["purchase_amount"],
    errors="coerce"
)
print(df["purchase_amount"])
print("\nData type:", df["purchase_amount"].dtype)
```

```python
Q1_age = df["age"].quantile(0.25)
Q3_age = df["age"].quantile(0.75)
IQR_age = Q3_age - Q1_age

lower_age = Q1_age - 1.5 * IQR_age
upper_age = Q3_age + 1.5 * IQR_age

age_outliers = df[
    (df["age"] < lower_age) |
    (df["age"] > upper_age)
]
print("Q1:", Q1_age)
print("Q3:", Q3_age)
print("IQR:", IQR_age)
print("Lower limit:", lower_age)
print("Upper limit:", upper_age)
print("\nAge outliers:")
print(age_outliers[["customer_id", "age"]])
```

## Outlier Treatment – Age

The IQR method identified one outlier in the `age` column:

- Customer `C011` has an age of `999`.
- The calculated upper IQR limit is `55`.
- Therefore, `999` is an extreme and unrealistic value.

### Decision
The value `999` will not be retained because it is clearly invalid for an age.

Instead of deleting the complete customer record, the invalid age will be replaced with a missing value and then imputed using the **median age (33)**.

This approach preserves the customer's other valid information while correcting the invalid age entry.

```python
df.loc[df["age"] == 999, "age"] = np.nan
df["age"] = df["age"].fillna(median_age)
print("Age value for C011:", df.loc[df["customer_id"] == "C011", "age"].iloc[0])
```

```python
Q1_amount = df["purchase_amount"].quantile(0.25)
Q3_amount = df["purchase_amount"].quantile(0.75)
IQR_amount = Q3_amount - Q1_amount
lower_amount = Q1_amount - 1.5 * IQR_amount
upper_amount = Q3_amount + 1.5 * IQR_amount

amount_outliers = df[
    (df["purchase_amount"] < lower_amount) |
    (df["purchase_amount"] > upper_amount)
]

print("Q1:", Q1_amount)
print("Q3:", Q3_amount)
print("IQR:", IQR_amount)
print("Lower limit:", lower_amount)
print("Upper limit:", upper_amount)
print("\nPurchase amount outliers:")
print(amount_outliers[["customer_id", "purchase_amount"]])
```

## Outlier Treatment – Purchase Amount

The IQR method identified four potential outliers in the `purchase_amount` column. The calculated acceptable range was from 200.25 to 1880.25.

- **C004 – 2100:** Retained because it is a plausible positive purchase amount.
- **C013 – 2300:** Retained because it is a plausible positive purchase amount.
- **C012 – -450:** Treated as invalid because a negative purchase amount is not reasonable for this dataset.
- **C017 – 15000:** Treated as an extreme anomaly because it is substantially higher than the other transaction amounts.

The two invalid/anomalous values will be replaced with missing values and imputed using the median purchase amount. This preserves the customer records while correcting unreasonable transaction values.

```python
df.loc[df["purchase_amount"] < 0, "purchase_amount"] = np.nan
df.loc[df["purchase_amount"] == 15000, "purchase_amount"] = np.nan

median_amount = df["purchase_amount"].median()
print("Median purchase amount:", median_amount)
```

```python
df["purchase_amount"] = df["purchase_amount"].fillna(median_amount)
print(
    "Missing purchase amount values after imputation:",
    df["purchase_amount"].isnull().sum()
)
```

```python
df["city"] = df["city"].str.strip().str.title()
df["city"] = df["city"].replace({
    "Bengaluru": "Bangalore"
})

print(df["city"].value_counts())
```

```python
original_dates = pd.read_csv("messy_retail_customers.csv")["purchase_date"]
```

```python
df["purchase_date"] = pd.to_datetime(
    original_dates,
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print(df["purchase_date"].head(10))

print("\nMissing dates after correction:",
      df["purchase_date"].isnull().sum())

print("Data type:", df["purchase_date"].dtype)
```

```python
print(df.dtypes)
```

```python
text_columns = [
    "customer_id",
    "customer_name",
    "gender",
    "city",
    "payment_method"
]

for col in text_columns:
    df[col] = df[col].astype("string")

df["age"] = df["age"].astype(float)
df["purchase_amount"] = df["purchase_amount"].astype(float)

print(df.dtypes)
```

```python
# Before vs. After cleaning summary

original_df = pd.read_csv("messy_retail_customers.csv")

dtype_columns_to_check = {
    "age": pd.api.types.is_numeric_dtype,
    "purchase_date": pd.api.types.is_datetime64_any_dtype,
    "purchase_amount": pd.api.types.is_numeric_dtype
}

before_dtype_issues = sum(
    not check(original_df[column].dtype)
    for column, check in dtype_columns_to_check.items()
)

# Count data type issues after cleaning
after_dtype_issues = sum(
    not check(df[column].dtype)
    for column, check in dtype_columns_to_check.items()
)

before_nulls = original_df.isnull().sum().sum()
before_duplicates = original_df.duplicated().sum()
before_rows = len(original_df)

after_nulls = df.isnull().sum().sum()
after_duplicates = df.duplicated().sum()
after_rows = len(df)

summary = pd.DataFrame({
    "Metric": [
        "Total Null Values",
        "Duplicate Rows",
        "Row Count",
        "Dtype Issues"
    ],
    "Before Cleaning": [
        before_nulls,
        before_duplicates,
        before_rows,
        before_dtype_issues
    ],
    "After Cleaning": [
        after_nulls,
        after_duplicates,
        after_rows,
        after_dtype_issues
    ]
})

summary
```
